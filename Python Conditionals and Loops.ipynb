```python
#Part A – Student Performance Checker 

student_name= input( "Enter student's name:")

# Last two digits of registration number 677909
subject1 = 9
print("Marks for subject 1:", subject1)

subject2= float( input( "Enter marks for subject 2:"))
subject3= float( input( "Enter marks for subject 3:"))

if subject1< 0 or subject1> 100 or subject2< 0 or subject2> 100 or subject3< 0 or subject3> 100:
    print( "Rejected. Only accept marks 0-100.")
else:
    total= subject1+subject2+subject3
    average= total/3

    if average>= 80:
        grade= "A"
        comment= "Excellent"
    elif average>= 70:
        grade= "B"
        comment= "Very Good"
    elif average>= 60:
        grade= "C"
        comment= "Good"
    elif average>= 50:
        grade= "D"
        comment= "Pass"
    else:
        grade= "F"
        comment= "Fail"

    print( "Student Name:", student_name)
    print( "Total Marks:", total)
    print( "Average Marks:", round( average, 2))
    print( "Grade:", grade)
    print( "Performance Comment:", comment)
```

```python
#Part B – Scholarship Eligibility Checker

age= int(input( "Enter student's age:"))
average= float( input( "Enter average marks:"))
registration_status= input( "Registered full-time? (yes/no):")
fee_balance= input( "Any outstanding fee balance? (yes/no):")

if age>= 18:
    if average>= 75:
        if registration_status== "yes":
            if fee_balance== "no":
                print( "Eligible for a scholarship.")
                print( "You are eligible.")
            else:
                print( "Not eligible: Outstanding fee balance.")
        else:
            print( "Not eligible: Full-time registration needed.")
    else:
        print( "Not eligible: Average marks must be at least 75.")
else:
    print( "Not eligible: You must be at least 18 years old.")
```

The additional condition is that the student needs not to have an outstanding fee balance so as to qualify to get a scholarship.

```python
#Part D – Shopping Basket Calculator

number_of_products= 12
total= 0
expensive_products= 0
prices= []

for i in range(1, number_of_products+1):
    price= float( input( "Enter price for product " + str(i) + ": Ksh "))

    while price in prices or price<= 0:
        print( "Give a positive price different from previous products.")
        price= float( input( "Enter price for product " + str(i) + ": Ksh "))

    prices.append( price)
    total= total+price

    if price> 1000:
        expensive_products= expensive_products+1

if total< 2000:
    discount_rate= 0
elif total< 5000:
    discount_rate= 0.05
elif total< 10000:
    discount_rate= 0.10
else:
    discount_rate= 0.15

discount_amount= total*discount_rate
final_amount= total-discount_amount

print( "Total purchase: Ksh", round( total, 2))
print( "Products costing more than Ksh 1,000:", expensive_products)
print( "Discount: Ksh", round( discount_amount, 2))
print( "Final payable amount: Ksh", round( final_amount, 2))
```

```python
#Part E – Simple ATM Simulator

balance= 10000
first_deposit= 9090
deposit_count= 0

while True:
    print( "\n===== SIMPLE ATM =====")
    print( "1. Check Balance")
    print( "2. Deposit Money")
    print( "3. Withdraw Money")
    print( "4. Exit")

    option= input( "Enter your choice: ")

    if option== "1":
        print( "Current balance: Ksh", balance)

    elif option== "2":
        if deposit_count== 0:
            amount= first_deposit
            print( "First deposit: Ksh", amount)
        else:
            amount= float( input( "Provide deposit amount: Ksh "))

        if amount<= 0:
            print( "Rejected. Amount must be greater than 0.")
        else:
            balance= balance+amount
            deposit_count= deposit_count+1
            print( "Successful deposit.")
            print( "New balance: Ksh", balance)

    elif option== "3":
        amount= float( input( "Provide withdrawal amount: Ksh "))

        if amount<= 0:
            print( "Rejected. Amount must be greater than 0.")
        elif amount> balance:
            print( "Rejected. Insufficient balance.")
        else:
            balance= balance-amount
            print( "Successful withdrawal.")
            print( "New balance: Ksh", balance)

    elif option== "4":
        print( "Success!")
        break

    else:
        print( "Invalid. Please select 1, 2, 3, or 4.")
```

```python
#Part F – Break and Continue

value= 0

while True:
    number= int(input( "Input a number: "))

    if number< 0:
        continue
    elif number== 0:
        break
    else:
        print( "Square:", number*number)
        value= value+1

print( "Number of positive numbers processed:", value)
```

```python
#Part G – Savings Tracker

savings= 3400

print( "Starting amount: Ksh", savings)

while True:
    amount= float(input( "Input amount to be saved: Ksh "))

    if amount== 0:
        break
    elif amount< 0:
        print( "Rejected. Amount must be positive.")
        continue
    else:
        savings= savings+amount
        print( "Total amount saved: Ksh", savings)

print( "Final amount saved: Ksh", savings)
```

The tracker allows for money to be added to one's savings. Its starting amount is Ksh 3400 and it stops when 0 is entered as the amount to be saved.
